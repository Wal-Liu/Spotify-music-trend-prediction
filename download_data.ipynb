```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("asaniczka/top-spotify-songs-in-73-countries-daily-updated")

print("Path to dataset files:", path)
```

```python
# Spotify path
spotify_path = f'{path}/universal_top_spotify_songs.csv'
print("Path to Spotify dataset file:", spotify_path)
```

```python
import pandas as pd
spotify_df = pd.read_csv(spotify_path)
```

```python
spotify_df.head()
```

```python
spotify_df = spotify_df[spotify_df['snapshot_date'] >= '2025-01-01']
```

```python
print("Filtered dataset shape:", spotify_df.shape)
```

```python
import os

os.makedirs('./data', exist_ok=True)
```

```python
# Save the filtered dataset to a new CSV file
spotify_df.to_csv('./data/2025_spotify_songs.csv', index=False)
```
