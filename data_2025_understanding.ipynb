```python
import pandas as pd
```

```python

data_2025 = pd.read_csv('./data/2025_spotify_songs.csv')
```

```python
# Kiểm tra thống kê mô tả cơ bản
print("=== THỐNG KÊ MÔ TẢ CƠ BẢN ===")
data_2025.describe()
```

```python
# Kiểm tra giá trị null và thống kê missing data
print("=== THỐNG KÊ MISSING DATA ===")
missing_data = data_2025.isnull().sum()
missing_percentage = (missing_data / len(data_2025)) * 100
missing_summary = pd.DataFrame({
    'Missing Count': missing_data,
    'Percentage': missing_percentage
}).sort_values('Missing Count', ascending=False)

print(missing_summary[missing_summary['Missing Count'] > 0])
```

The `NaN` values ​​in the `country` column are not data errors, but come from Spotify's **Global Top 50** playlist.
According to the official documentation of the **Spotify Charts API**:
> “If NULL, then the playlist is ‘Global’. Since Global doesn’t have an ISO code, it is not put here.”

This means that records without a country code (`NULL` or `NaN`) are songs in the **global chart**.
Therefore, we will replace the missing value with `'Global'` to make this clearer:

```python
data_2025['country'] = data_2025['country'].fillna('Global')
```

```python
data_2025 = data_2025.dropna(subset=['album_name', 'album_release_date', 'artists', 'name'])
```

The columns album_name, album_release_date, artists, and name contain a very small number of missing values (less than 0.001% of total records).
Since these records have negligible impact on the analysis, the missing values were handled by replacing them with descriptive placeholders such as “Unknown” or by removing the affected rows to maintain data consistency.

```python
# Cài đặt các thư viện cần thiết
%pip install seaborn matplotlib numpy

# Import các thư viện cần thiết cho visualization
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime

# Thiết lập style cho plots
plt.style.use('default')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 10
```

```python
# Phân tích phân phối của các đặc trưng âm nhạc (Audio Features)
print("=== PHÂN TÍCH CÁC ĐẶC TRƯNG ÂM NHẠC ===")

# Danh sách các audio features
audio_features = ['danceability', 'energy', 'speechiness', 'acousticness', 
                 'instrumentalness', 'liveness', 'valence', 'loudness', 'tempo']

# Tạo subplot cho các audio features
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.ravel()

for i, feature in enumerate(audio_features):
    axes[i].hist(data_2025[feature], bins=50, alpha=0.7, edgecolor='black')
    axes[i].set_title(f'Phân phối của {feature}')
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel('Tần suất')

plt.tight_layout()
plt.show()
```

```python
# Phân tích ma trận tương quan giữa các audio features
print("=== MA TRẬN TƯƠNG QUAN CÁC ĐẶC TRƯNG ÂM NHẠC ===")

correlation_features = ['danceability', 'energy', 'speechiness', 'acousticness', 
                       'instrumentalness', 'liveness', 'valence', 'loudness', 'tempo', 'popularity']

correlation_matrix = data_2025[correlation_features].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=0.5)
plt.title('Ma trận tương quan giữa các đặc trưng âm nhạc và popularity')
plt.show()
```

```python
# Phân tích ranking và popularity
print("=== PHÂN TÍCH RANKING VÀ POPULARITY ===")

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Phân phối daily_rank
axes[0,0].hist(data_2025['daily_rank'], bins=50, alpha=0.7, edgecolor='black')
axes[0,0].set_title('Phân phối Daily Rank')
axes[0,0].set_xlabel('Daily Rank')
axes[0,0].set_ylabel('Tần suất')

# 2. Phân phối popularity
axes[0,1].hist(data_2025['popularity'], bins=50, alpha=0.7, edgecolor='black')
axes[0,1].set_title('Phân phối Popularity')
axes[0,1].set_xlabel('Popularity')
axes[0,1].set_ylabel('Tần suất')

# 3. Scatter plot giữa daily_rank và popularity
sample_data = data_2025.sample(n=10000)  # Lấy mẫu để vẽ nhanh hơn
axes[1,0].scatter(sample_data['daily_rank'], sample_data['popularity'], alpha=0.5)
axes[1,0].set_title('Mối quan hệ giữa Daily Rank và Popularity')
axes[1,0].set_xlabel('Daily Rank')
axes[1,0].set_ylabel('Popularity')

# 4. Box plot popularity theo các nhóm rank
data_2025['rank_group'] = pd.cut(data_2025['daily_rank'], 
                                bins=[0, 10, 50, 100, 200, 1000], 
                                labels=['Top 10', '11-50', '51-100', '101-200', '200+'])
axes[1,1].boxplot([data_2025[data_2025['rank_group'] == group]['popularity'].values 
                   for group in ['Top 10', '11-50', '51-100', '101-200', '200+']])
axes[1,1].set_title('Popularity theo nhóm Ranking')
axes[1,1].set_xlabel('Nhóm Ranking')
axes[1,1].set_ylabel('Popularity')
axes[1,1].set_xticklabels(['Top 10', '11-50', '51-100', '101-200', '200+'])

plt.tight_layout()
plt.show()
```

```python
# Phân tích theo quốc gia
print("=== PHÂN TÍCH THEO QUỐC GIA ===")

# Top 10 quốc gia có nhiều bài hát nhất
country_counts = data_2025['country'].value_counts().head(10)
print("Top 10 quốc gia có nhiều bài hát nhất trong dataset:")
print(country_counts)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 1. Bar chart top countries
axes[0].bar(range(len(country_counts)), country_counts.values)
axes[0].set_title('Top 10 quốc gia có nhiều bài hát nhất')
axes[0].set_xlabel('Quốc gia')
axes[0].set_ylabel('Số lượng bài hát')
axes[0].set_xticks(range(len(country_counts)))
axes[0].set_xticklabels(country_counts.index, rotation=45)

# 2. Popularity trung bình theo top countries
top_countries = country_counts.head(10).index
avg_popularity_by_country = data_2025[data_2025['country'].isin(top_countries)].groupby('country')['popularity'].mean().sort_values(ascending=False)

axes[1].bar(range(len(avg_popularity_by_country)), avg_popularity_by_country.values)
axes[1].set_title('Popularity trung bình theo quốc gia (Top 10)')
axes[1].set_xlabel('Quốc gia')
axes[1].set_ylabel('Popularity trung bình')
axes[1].set_xticks(range(len(avg_popularity_by_country)))
axes[1].set_xticklabels(avg_popularity_by_country.index, rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Phân tích thời gian (snapshot_date)
print("=== PHÂN TÍCH THEO THỜI GIAN ===")

# Chuyển đổi snapshot_date sang datetime
data_2025['snapshot_date'] = pd.to_datetime(data_2025['snapshot_date'])

# Phân tích theo ngày/tháng
data_2025['month'] = data_2025['snapshot_date'].dt.month
data_2025['day_of_week'] = data_2025['snapshot_date'].dt.day_name()

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Số lượng bài hát theo tháng
monthly_counts = data_2025['month'].value_counts().sort_index()
axes[0,0].plot(monthly_counts.index, monthly_counts.values, marker='o')
axes[0,0].set_title('Số lượng bài hát theo tháng')
axes[0,0].set_xlabel('Tháng')
axes[0,0].set_ylabel('Số lượng bài hát')
axes[0,0].set_xticks(range(1, 13))

# 2. Popularity trung bình theo tháng
monthly_popularity = data_2025.groupby('month')['popularity'].mean()
axes[0,1].plot(monthly_popularity.index, monthly_popularity.values, marker='o', color='orange')
axes[0,1].set_title('Popularity trung bình theo tháng')
axes[0,1].set_xlabel('Tháng')
axes[0,1].set_ylabel('Popularity trung bình')
axes[0,1].set_xticks(range(1, 13))

# 3. Số lượng bài hát theo ngày trong tuần
daily_counts = data_2025['day_of_week'].value_counts()
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_counts = daily_counts.reindex(day_order)
axes[1,0].bar(range(len(daily_counts)), daily_counts.values)
axes[1,0].set_title('Số lượng bài hát theo ngày trong tuần')
axes[1,0].set_xlabel('Ngày trong tuần')
axes[1,0].set_ylabel('Số lượng bài hát')
axes[1,0].set_xticks(range(len(daily_counts)))
axes[1,0].set_xticklabels([day[:3] for day in daily_counts.index], rotation=45)

# 4. Timeline of data collection
date_counts = data_2025['snapshot_date'].value_counts().sort_index()
axes[1,1].plot(date_counts.index, date_counts.values)
axes[1,1].set_title('Timeline thu thập dữ liệu')
axes[1,1].set_xlabel('Ngày')
axes[1,1].set_ylabel('Số lượng records')
axes[1,1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Phân tích nghệ sĩ và bài hát phổ biến
print("=== PHÂN TÍCH NGHỆ SĨ VÀ BÀI HÁT PHỔ BIẾN ===")

# Top 10 bài hát phổ biến nhất (theo popularity trung bình)
top_songs = data_2025.groupby(['name', 'artists'])['popularity'].mean().reset_index().sort_values('popularity', ascending=False).head(10)
print("Top 10 bài hát có popularity cao nhất:")
for i, row in top_songs.iterrows():
    print(f"{i+1}. {row['name']} - {row['artists']} (Popularity: {row['popularity']:.1f})")

print("\n" + "="*50 + "\n")

# Top 10 nghệ sĩ có nhiều bài hát nhất trong dataset
top_artists = data_2025['artists'].value_counts().head(10)
print("Top 10 nghệ sĩ có nhiều bài hát nhất trong dataset:")
print(top_artists)

# Visualization
fig, axes = plt.subplots(2, 1, figsize=(15, 12))

# 1. Top songs by popularity
top_songs_plot = top_songs.head(10)
song_labels = [f"{row['name'][:20]}..." if len(row['name']) > 20 else row['name'] for _, row in top_songs_plot.iterrows()]
axes[0].barh(range(len(top_songs_plot)), top_songs_plot['popularity'])
axes[0].set_title('Top 10 bài hát có popularity cao nhất')
axes[0].set_xlabel('Popularity')
axes[0].set_ylabel('Bài hát')
axes[0].set_yticks(range(len(top_songs_plot)))
axes[0].set_yticklabels(song_labels)

# 2. Top artists by song count
axes[1].barh(range(len(top_artists)), top_artists.values)
axes[1].set_title('Top 10 nghệ sĩ có nhiều bài hát nhất')
axes[1].set_xlabel('Số lượng bài hát')
axes[1].set_ylabel('Nghệ sĩ')
axes[1].set_yticks(range(len(top_artists)))
artist_labels = [artist[:30] + "..." if len(artist) > 30 else artist for artist in top_artists.index]
axes[1].set_yticklabels(artist_labels)

plt.tight_layout()
plt.show()
```

```python
# Phân tích âm sắc và đặc điểm âm nhạc
print("=== PHÂN TÍCH ÂM SẮC VÀ ĐẶC ĐIỂM ÂM NHẠC ===")

# Phân tích key và mode
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Phân phối Key (Tone)
key_counts = data_2025['key'].value_counts().sort_index()
key_names = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']
axes[0,0].bar(range(len(key_counts)), key_counts.values)
axes[0,0].set_title('Phân phối Key (Tone) trong âm nhạc')
axes[0,0].set_xlabel('Key')
axes[0,0].set_ylabel('Số lượng bài hát')
axes[0,0].set_xticks(range(len(key_counts)))
axes[0,0].set_xticklabels(key_names)

# 2. Phân phối Mode (Major/Minor)
mode_counts = data_2025['mode'].value_counts()
mode_labels = ['Minor', 'Major']
axes[0,1].pie(mode_counts.values, labels=mode_labels, autopct='%1.1f%%')
axes[0,1].set_title('Phân phối Mode (Major/Minor)')

# 3. Phân phối Time Signature
time_sig_counts = data_2025['time_signature'].value_counts().sort_index()
axes[1,0].bar(range(len(time_sig_counts)), time_sig_counts.values)
axes[1,0].set_title('Phân phối Time Signature')
axes[1,0].set_xlabel('Time Signature')
axes[1,0].set_ylabel('Số lượng bài hát')
axes[1,0].set_xticks(range(len(time_sig_counts)))
axes[1,0].set_xticklabels(time_sig_counts.index)

# 4. Phân phối Duration
data_2025['duration_minutes'] = data_2025['duration_ms'] / 60000
axes[1,1].hist(data_2025['duration_minutes'], bins=50, alpha=0.7, edgecolor='black')
axes[1,1].set_title('Phân phối thời lượng bài hát (phút)')
axes[1,1].set_xlabel('Thời lượng (phút)')
axes[1,1].set_ylabel('Tần suất')

plt.tight_layout()
plt.show()

# Thống kê cơ bản về thời lượng
print(f"Thời lượng trung bình: {data_2025['duration_minutes'].mean():.2f} phút")
print(f"Thời lượng ngắn nhất: {data_2025['duration_minutes'].min():.2f} phút")
print(f"Thời lượng dài nhất: {data_2025['duration_minutes'].max():.2f} phút")
print(f"Thời lượng median: {data_2025['duration_minutes'].median():.2f} phút")
```

```python
# Phân tích nội dung explicit và album
print("=== PHÂN TÍCH NỘI DUNG EXPLICIT VÀ ALBUM ===")

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Tỷ lệ explicit content
explicit_counts = data_2025['is_explicit'].value_counts()
axes[0,0].pie(explicit_counts.values, labels=['Non-Explicit', 'Explicit'], autopct='%1.1f%%')
axes[0,0].set_title('Tỷ lệ Explicit Content')

# 2. Popularity comparison: Explicit vs Non-Explicit
explicit_popularity = data_2025.groupby('is_explicit')['popularity'].mean()
axes[0,1].bar(['Non-Explicit', 'Explicit'], explicit_popularity.values)
axes[0,1].set_title('Popularity trung bình: Explicit vs Non-Explicit')
axes[0,1].set_ylabel('Popularity trung bình')

# 3. Phân tích album release date
data_2025['album_release_date'] = pd.to_datetime(data_2025['album_release_date'], errors='coerce')
data_2025['release_year'] = data_2025['album_release_date'].dt.year

# Lọc các năm hợp lý (1950-2025)
valid_years = data_2025[(data_2025['release_year'] >= 1950) & (data_2025['release_year'] <= 2025)]
year_counts = valid_years['release_year'].value_counts().sort_index()

# Chỉ hiển thị từ 2000 trở đi để chart dễ nhìn
recent_years = year_counts[year_counts.index >= 2000]
axes[1,0].plot(recent_years.index, recent_years.values, marker='o')
axes[1,0].set_title('Số lượng bài hát theo năm phát hành (2000+)')
axes[1,0].set_xlabel('Năm phát hành')
axes[1,0].set_ylabel('Số lượng bài hát')
axes[1,0].tick_params(axis='x', rotation=45)

# 4. Top 10 albums phổ biến nhất
top_albums = data_2025['album_name'].value_counts().head(10)
album_labels = [album[:25] + "..." if len(album) > 25 else album for album in top_albums.index]
axes[1,1].barh(range(len(top_albums)), top_albums.values)
axes[1,1].set_title('Top 10 Albums có nhiều bài hát nhất')
axes[1,1].set_xlabel('Số lượng bài hát')
axes[1,1].set_ylabel('Album')
axes[1,1].set_yticks(range(len(top_albums)))
axes[1,1].set_yticklabels(album_labels)

plt.tight_layout()
plt.show()

print(f"Tổng số album unique: {data_2025['album_name'].nunique()}")
```
