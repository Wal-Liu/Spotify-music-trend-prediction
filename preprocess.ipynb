# Preprocess the data

```python
import pandas as pd
spotify_path = './data/2025_spotify_songs.csv'
```

```python
songs = pd.read_csv(spotify_path)
```

```python
songs.head()
```

```python
songs.info()
```

```python
# Get songs with non-null name
clean_songs = songs[~songs['name'].isnull()]
```

```python
clean_songs.info()
```

```python
clean_songs = clean_songs[clean_songs['country'].notna()]
clean_songs = clean_songs[clean_songs['album_name'].notna()]
```

```python
clean_songs.info()
```

```python
# Normalize data types

# Convert date columns to datetime type
clean_songs['snapshot_date'] = pd.to_datetime(clean_songs['snapshot_date'], errors='coerce')
clean_songs['album_release_date'] = pd.to_datetime(clean_songs['album_release_date'], errors='coerce')

# Convert integer columns to int type
clean_songs['daily_rank'] = clean_songs['daily_rank'].astype(int)
clean_songs['daily_movement'] = clean_songs['daily_movement'].astype(int)
clean_songs['weekly_movement'] = clean_songs['weekly_movement'].astype(int)
clean_songs['popularity'] = clean_songs['popularity'].astype(int)
clean_songs['duration_ms'] = clean_songs['duration_ms'].astype(int)
clean_songs['key'] = clean_songs['key'].astype(int)
clean_songs['mode'] = clean_songs['mode'].astype(int)
clean_songs['time_signature'] = clean_songs['time_signature'].astype(int)

# Ensure float columns stay in float type
float_columns = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 
                 'instrumentalness', 'liveness', 'valence', 'tempo']
clean_songs[float_columns] = clean_songs[float_columns].astype(float)

# Clean string columns (strip whitespace, convert to lowercase if needed)
clean_songs['spotify_id'] = clean_songs['spotify_id'].str.strip()
clean_songs['name'] = clean_songs['name'].str.strip()
clean_songs['artists'] = clean_songs['artists'].str.strip()
clean_songs['country'] = clean_songs['country'].str.strip()
clean_songs['album_name'] = clean_songs['album_name'].str.strip()
```

```python
clean_songs = clean_songs.drop_duplicates()
```

```python
# Save the clean dataset to a new CSV file
clean_songs.to_csv('./data/2025_spotify_songs_cleaned.csv', index=False)
```
